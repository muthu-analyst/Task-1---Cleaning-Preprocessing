```python
# 2. standardize column names
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_').str.replace(r'[^0-9a-z_]', '', regex=True)
```

```python
# 5. duplicates
df.drop_duplicates(inplace=True)
```

```python
# Check for missing values
print(df.isnull().sum())
```

```python

# For numerical columns, you could fill NAs with median or mean (example: 'income')
df['income'] = df['income'].fillna(df['income'].median())

# For categorical columns, fill NAs with most frequent value
df['education'] = df['education'].fillna(df['education'].mode()[0])
```

```python
# Convert date columns to datetime (example: 'dt_customer')
df['dt_customer'] = pd.to_datetime(df['dt_customer'])

# Convert categorical variables
categorical_columns = ['education', 'marital_status', 'country']
for col in categorical_columns:
    df[col] = df[col].astype('category')
```

```python
# 'income', clamp values between 1st and 99th percentiles
income_q01 = df['income'].quantile(0.01)
income_q99 = df['income'].quantile(0.99)
df = df[(df['income'] >= income_q01) & (df['income'] <= income_q99)]
```

```python
df['dt_customer'] = pd.to_datetime(df['dt_customer'], errors='coerce')  # Convert to datetime
df['dt_customer'] = df['dt_customer'].dt.strftime('%d-%m-%Y')           # Format as dd-mm-yyyy string
```
